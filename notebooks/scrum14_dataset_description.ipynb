```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', font_scale=1.1)
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['figure.dpi'] = 100

DATA_DIR = os.path.join('..', 'data')
PROCESSED_DIR = os.path.join(DATA_DIR, 'processed')
DOCS_DIR = os.path.join('..', 'docs')
FIGURES_DIR = os.path.join(DOCS_DIR, 'figures')
print('Setup complete.')
```

## 1. Product Registry Portfolio Composition (N=60)

Analyzing the distribution of the 60 smartphone models across brands and lifecycle cohorts.

```python
products = pd.read_csv(os.path.join(DATA_DIR, 'products_v3.csv'))
print(f'Total Products: {len(products)}')
cohort_summary = products['product_type'].value_counts().reset_index()
cohort_summary.columns = ['Cohort', 'Count']
cohort_summary['Share'] = (cohort_summary['Count'] / len(products)).map('{:.1%}'.format)
print('--- Product Lifecycle Cohorts ---')
print(cohort_summary.to_string(index=False))
```

```python
brand_dist = products.groupby(['brand', 'product_type']).size().unstack(fill_value=0)
brand_dist['Total'] = brand_dist.sum(axis=1)
brand_dist = brand_dist.sort_values(by='Total', ascending=False)
print('--- Brand by Lifecycle Stage Matrix ---')
print(brand_dist)
```

## 2. Multi-Source Signal Volume Metrics

Compiling volume, coverage, and granularity metrics for Google Trends and YouTube Data API.

```python
trends = pd.read_csv(os.path.join(PROCESSED_DIR, 'trends_clean.csv'))
videos = pd.read_csv(os.path.join(PROCESSED_DIR, 'youtube_videos_clean.csv'))
comments = pd.read_csv(os.path.join(PROCESSED_DIR, 'youtube_comments_clean.csv'))
aligned = pd.read_csv(os.path.join(PROCESSED_DIR, 'aligned_daily.csv'))

metrics = pd.DataFrame([
    {'Dimension': 'Product Scope', 'Metric': f"{products['product_id'].nunique()} Smartphones", 'Scope': '13 Brands (India)'},
    {'Dimension': 'Google Trends Observations', 'Metric': f'{len(trends):,} Daily Rows', 'Scope': '269 Days continuous'},
    {'Dimension': 'YouTube Videos Extracted', 'Metric': f'{len(videos):,} Videos', 'Scope': 'Top 50 Relevance search'},
    {'Dimension': 'YouTube Comments Parsed', 'Metric': f'{len(comments):,} Comments', 'Scope': 'Top 5 Videos x 100 comments'},
    {'Dimension': 'Unified Aligned Daily Index', 'Metric': f'{len(aligned):,} Rows', 'Scope': 'Multi-source joint table'}
])
print('--- Slide Summary Metrics Table ---')
print(metrics.to_string(index=False))
```

## 3. Data Quality, Usability & Exact-Model Matching

Evaluating precision thresholds (`on_target_share >= 0.60`) and usable product yield.

```python
report = pd.read_csv(os.path.join(PROCESSED_DIR, 'full_report.csv'))
usable_count = int(report['usable'].sum())
low_signal_count = int((report['low_signal_product'] == True).sum())
low_target_count = int((report['low_on_target_share'] == True).sum())

print(f'Total Registered Products: {len(report)}')
print(f'Usable on Both Signals: {usable_count} ({usable_count / len(report):.1%})')
print(f'Zero-Heavy Trends Products (>50% zero): {low_signal_count}')
print(f'Low On-Target YouTube Share (<60% precision): {low_target_count}')
```
